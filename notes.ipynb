Individual roll command called Starlight Life Apocalypse which has varies options
The apocalypse will essentially be every normal person has the role "Light" which just has the description "You are normal! :)" and alternatively there is a roll called "The Darkness", the Darkness is effectivley a zombie varient whos goal is to convert all the other players into "Dark Matter", basically the Darkness has to convert all the other players into Dark Matter otherwise they will be put down 2 life stages at the end of the session, there will be 2 Darkness one being pre-set (so a command to set a darkness before its rolled) and a second completely random one (you are told who it is in the role description), when they kill someone that player turns into "Dark Matter" and there goal is to spread the infection otherwise they will go down a life stage (when they go down stages they are automatically set to 100 starlight). The Darkness/Dark Matter succeed if every player is converted into Dark Matter and the Light succeed if they survive until the end of session, if anyone is Light by the end of session the darkness/darkmatter receive their collective punishments (Dark Matter one life stage, Darkness two stages).

Further Details
Command List
/StarlightLife Apocalypse Roll - Rolls the Light and the Darkness
/StarlightLife Apocalypse PreSet - Sets the one predetermined Darkness
/StarlightLife Apocalypse SetRole - Sets players as Dark Matter if they are killed indirectly and allows for players to be set back to light if they die illegally
/StarlightLife Apocalypse DarkCure - Cures all the Darkness/DarkMatter
/StarlightLife Apocalypse DarkFail - Fails all Dark Roles and clears the Light of their roles
/StarlightLife Apocolypse End - Ends the apocalypse, failsafe incase something goes wrong and automatically cures/fails roles.
/StarlightLife Apocolypse LightGlow - Gives all Light Players permanent glowing
/Checkrole - public command, tells what your role is and rereads role description

Light
Role description
"You are Light! You are normal!"
Once 10 Players are Darkness/Dark Matter, every 5 minutes all Light Players will receive Glowing for 30 seconds
They are cleared of being light at the end of session
Forwarded
Dark Matter
Role Description
"You are Dark Matter, you have been infected and must work with other Dark Matter as well as the Darkness infect every other player via killing them direct and indirect, all teams are void while you are Dark Matter and if you fail to infect every Light player you will be set down to the next Life Color, you may kill other Dark Matter however this will not help your overall goal."
Dark Matter every 20 minutes will recieve darkness for 15 seconds and lose 1-3 Starlight
Once 6 people are infected (Dark Matter), all Dark Matters names will appear Dark Gray on the tablist and the Two Darkness will appear purple (They will still have their individual colors and stages the new colors will just show above their name and in tab/chat"
They are cleared of being Dark Matter if all light players are converted into Dark Matter (Success), or if they fail to infect Everyone (Fail, -1 life stage set to 100 starlight)

The Darkness
Role Description
"You are the Darkness, all of your teams are invalid and your goal is now to infect all the other light players and convert them into Dark Matter via killing them direct or indirect, you act as a leader to the Dark Matter as you are the root of the infection, therefore if you die all existing Dark Matter will also lose 10 starlight. if you fail to convert every other player into Dark Matter you will be set down 2 Life Colors, you may kill other Dark Matter and Darkness players if you wish however this will not contribute towards your overall goal. [INSERT USERNAME] Is the other Darkness, you may work with them to infect the rest of the players"
Dark Matter will recieve permanent Speed 1 once 12 Players are converted into Dark Matter
If they die all other dark matter players will also lose 10 starlight
They are cleared of Darkness if all light players are converted into Dark Matter (Success), or if they fail to infect Everyone (Fail, -2 Life Stages set to 100 Starlight)
